### Question 9
Which industry has the highest automation risk?

Right — in this case both .mean() and .max() happen to point to the same industry, Administrative & Clerical, just with different numbers attached (0.789742 vs 0.855). That's a bit of a coincidence with this dataset, not a rule you can rely on.
Here's why they're different numbers and why that matters:
.mean() result (0.789742) = the average automation risk score across all "Administrative & Clerical" rows in your data
.max() result (0.855) = the single highest score among all "Administrative & Clerical" rows — i.e., one particular record/year/company in that industry that scored 0.855
They agree on which industry here because Administrative & Clerical is probably consistently high-risk across most of its records — so it wins both "on average" and "at its peak." But this won't always happen. Imagine:
Industry A: scores are [0.9, 0.2, 0.2, 0.2] → mean = 0.375, max = 0.9
Industry B: scores are [0.6, 0.6, 0.6, 0.6] → mean = 0.6, max = 0.6
Here, .max() would say Industry A has the "highest automation risk" (because of one outlier record hitting 0.9), while .mean() would correctly say Industry B is consistently riskier overall (0.6 average vs 0.375 average).
So the takeaway: use .mean() when the question is about the industry as a whole ("which industry is riskiest overall"), and .max() only when you specifically want the single most extreme record. Since Question 9 asks about the industry, .mean() is the more correct and defensible choice — even though in your dataset it happened to give the same answer as .max().

### Question 27
AI Adoption by Region

```python
df.groupby('region')['ai_adoption_index'].mean().plot(kind='barh')
plt.title('AI Adoption by Region')
plt.xlabel('ai_adoption_index')
plt.ylabel('region')
plt.show()
```

## AI Adoption by Region — Analysis

# Statistical Insight**

# The regions Leading in AI adoption:
1. North America — highest, ~0.87
2. East Asia — ~0.85
3. Oceania — ~0.83
4. Europe — ~0.80
5. Middle East — ~0.72

# Lagging regions:
South Asia, Middle East/Africa, and Africa are the three lowest-adopting regions, all clustered around 0.53–0.58 — the bottom of the chart.

The spread across regions is meaningful — roughly 0.87 down to 0.53, a gap of over 30 percentage points between the highest and lowest adopting regions. This isn't a small variation; it's a clear divide between AI-mature regions and regions still catching up.

---

# Business Insight**

**Leading:** North America, East Asia, and Oceania are the clear frontrunners in AI adoption. These regions are typically associated with stronger digital infrastructure, larger tech investment budgets, and more mature enterprise software ecosystems — adoption tends to follow where the capital and infrastructure already exist.

**Lagging:** Africa, Middle East/Africa, and South Asia sit at the bottom of the chart. This likely reflects a mix of infrastructure gaps, lower capital availability, and possibly different regulatory or market readiness for large-scale AI deployment.

---

**Conclusion**

AI adoption is not evenly distributed globally — it's concentrated in a handful of economically developed regions (led by North America, East Asia, and Oceania), while a distinct group of regions (Africa, Middle East/Africa, South Asia) is still in earlier stages of adoption. This creates both a competitive gap and, from a business standpoint, an opportunity gap — the lagging regions represent largely untapped markets for AI-driven products and services.

---

**Business Recommendation**

- **For companies selling AI products/services:** Treat the leading regions — North America, East Asia, and Oceania — as mature markets. Competition here is likely higher, but so is buyer readiness, so focus on differentiation and advanced use cases rather than basic AI education.
- **For growth strategy:** The lagging regions — Africa, Middle East/Africa, and South Asia — represent whitespace opportunity: less competition, but also more groundwork needed (infrastructure partnerships, education, localized pricing) before adoption accelerates there.
- **For workforce and policy planning:** Given the earlier finding that high AI adoption correlates with workforce displacement risk, the leading regions should be the first priority for proactive workforce transition planning, since they're furthest along the adoption curve and most likely to hit the displacement "tipping point" first.

## Key Insight: AI Adoption and Workforce Displacement (2020–2026)
The analysis reveals a clear and consistent positive relationship between AI adoption and workforce displacement across the sector.
# Trend Analysis: 
Between 2020 and 2026, the AI adoption index rose steadily from approximately 0.55 to 0.85, while the percentage of the sector's workforce displaced increased from roughly 3% to 9% over the same period. Both metrics exhibit a similar upward trajectory, with growth accelerating notably after 2023.
# Relationship Analysis: 
The scatter plot analysis corroborates this finding at a more granular level — sectors with higher AI adoption indices consistently show higher rates of workforce displacement. Notably, the relationship shows diminishing marginal impact at higher adoption levels, suggesting a possible saturation point where further AI adoption yields comparatively smaller increases in displacement.
# Conclusion: 
Taken together, the time-series and cross-sectional data support the conclusion that increased AI adoption is associated with rising workforce displacement. Both metrics move in tandem over time, and the pattern holds when comparing across observations at a given point in time — strengthening confidence in the relationship.
Caveat: This analysis establishes correlation, not causation. Other confounding factors — such as broader economic cycles, sector-specific automation investment, labor market shifts, or policy changes — could be contributing to both trends simultaneously. A rigorous causal claim would require additional analysis (e.g., controlling for confounders, time-lag analysis to test whether adoption precedes displacement, or a difference-in-differences approach across sectors with varying adoption rates).

### Question 32
AI Adoption vs Workforce Displacement - Scatter Plot

```python
sns.scatterplot(data=df,x='ai_adoption_index',y='pct_sector_workforce_displaced',hue='pct_sector_workforce_displaced')
plt.title('AI Adoption vs Workforce Displacement - Scatter Plot')
plt.xlabel('ai_adoption_index')
plt.ylabel('pct_sector_workforce_displaced')
plt.show()
```

## AI Adoption vs. Workforce Displacement
The data shows a clear correlation between AI adoption and workforce displacement across sectors. As AI adoption increases, workforce displacement rises with it — but not in a straight line. Displacement stays modest through low and moderate adoption levels, then accelerates sharply once adoption crosses roughly 70-80%. Past that point, we're not just seeing more displacement — we're seeing it become less predictable, with a much wider range of outcomes across sectors.
That inflection point is the key takeaway. It gives us a real window to scale AI adoption while keeping workforce impact manageable, but that window narrows fast as adoption approaches saturation. The businesses that plan for the acceleration phase in advance — reskilling, redeployment, phased rollout — will be in a materially better position than those that treat displacement as a linear, predictable cost.
